```python
import os
import sys
from os import listdir
from os.path import isfile, join

import numpy as np
import pandas as pd
from termcolor import colored, cprint

from swamp.glob import glob

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path_input = glob.PATH_GENO + 'mini_dim/'
```

```python
PATH_humans = glob.PATH_PHENO_ancestry
humans = pd.read_csv(PATH_humans, sep="\t")
humans = humans.drop(columns=['Unnamed: 5', 'Unnamed: 6'])
humans = humans.rename({'sample': 'FID', 'sample.1': 'IID'}, axis=1)
humans['ancestry'] = humans['super_pop']
humans['ancestry'].mask(humans['super_pop'] == 'AFR', 0, inplace=True)
humans['ancestry'].mask(humans['super_pop'] == 'AMR', 1, inplace=True)
humans['ancestry'].mask(humans['super_pop'] == 'EAS', 2, inplace=True)
humans['ancestry'].mask(humans['super_pop'] == 'EUR', 3, inplace=True)
humans['ancestry'].mask(humans['super_pop'] == 'SAS', 4, inplace=True)
```

```python
humans['pop'].unique()
```

```python
import math
from sklearn import metrics
```

```python
def eval(path_file):
  # create array of different dimensions
  
  return metrics.silhouette_score(
      pd.read_pickle(path_file),
      humans['pop'],
      metric="euclidean"
  )
```

```python
onlyfiles_input = [f for f in listdir(path_input)]
```

```python
onlyfiles_shuffled = [f for f in onlyfiles_input if f.startswith('reduction_of_shuffled')]
onlyfiles_shuffled = [f for f in onlyfiles_shuffled if f.split('_')[3] != 'unshuffled']
```

```python
# check performance of shuffled
```

```python
eval_shuffled = []
for file in onlyfiles_shuffled:
  time = int(file.split("_")[5])
  path_file = path_input + file
  temp = eval(path_file)
  eval_shuffled.append([file.split("_")[3], time, temp, 'shuffled'])
```

```python
# Check performance of unshuffled
```

```python
onlyfiles_unshuffled = [f for f in onlyfiles_input if f.startswith('reduction_of_unshuffled')]
```

```python
eval_unshuffled = []
for file in onlyfiles_unshuffled:
  dim = "dim_" + file.split("_")[3]
  time = int(file.split("_")[5])
  temp = eval(path_file)
  path_file = path_input + file
  eval_unshuffled.append(
      [file.split("_")[3], time, temp, 'unshuffled'])
```

```python
# Check performance of shuffled unshuffled
```

```python
onlyfiles_shuffled_unshuffled = [
    f for f in onlyfiles_input if f.startswith('reduction_of_shuffled_unshuffled')]
```

```python
eval_shuffeld_unshuffled = []
for file in onlyfiles_shuffled_unshuffled:
  time = int(file.split("_")[6])
  path_file = path_input + file
  temp = eval(path_file)
  eval_shuffeld_unshuffled.append(
      [file.split("_")[4], time, temp, 'shuffled + unshuffled'])
```

```python
import seaborn as sns
```

```python
import pandas as pd
```

```python
shu_unshu = pd.DataFrame(data=eval_shuffeld_unshuffled)
shu_unshu = shu_unshu.rename(
    columns={0: "dim", 1: "time", 2: "eval", 3: "type"}, errors="raise")
```

```python
pd.DataFrame(data=eval_shuffeld_unshuffled)
```

```python
shu = pd.DataFrame(data=eval_shuffled)
shu = shu.rename(
    columns={0: "dim", 1: "time", 2: "eval", 3: "type"}, errors="raise")
```

```python
unshu = pd.DataFrame(data=eval_unshuffled)
unshu = unshu.rename(
    columns={0: "dim", 1: "time", 2: "eval", 3: "type"}, errors="raise")
```

```python
total = pd.concat([shu, unshu, shu_unshu],
                  ignore_index=True).sort_values(by=['dim'])
```

```python
palette = sns.color_palette("mako_r", 6)
```

```python
total['dim'] = total['dim'].astype(int)
```

```python
plt.figure(figsize = (15, 10))
sns.lineplot(
    data=total, x="dim", y="eval",
    hue="type",
    palette=palette
)
plt.savefig(f"silhouette.png")
plt.close()
```
